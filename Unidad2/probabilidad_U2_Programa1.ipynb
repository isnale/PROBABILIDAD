##Ejemplo 1: Nombres de personas

```python
import math as mt
from itertools import combinations
from itertools import combinations_with_replacement
import numpy as np
```

```python
import math as mt
from itertools import combinations

nombres = ["Oscar","Paco","Paty","Laura","Rubén","Luis","Lucy","Alberto","Juan","Danny"]
n = len(nombres)
r = 2
Cn = mt.factorial(n)/(mt.factorial(r)* mt.factorial(n-r))
print("Existen",Cn,"posibles combinaciones sin repitición del total de",n,"nombres", "en grupos de", r,"en",r)
print("Aquí las posibles combinaciones:")
combinaciones_nombres = list(combinations(nombres,r))
for combinations in combinaciones_nombres:
  print(combinations)
```

##Ejemplo 2 Caracteres alfabéticos y dígitos

```python
import math as mt
from itertools import combinations


caracteres = ["A","B","C","D","E","F","0","1","2","3","4","5"]
n = len(caracteres)
r = 10
Cn = mt.factorial(n)/(mt.factorial(r)* mt.factorial(n-r))
print("Existen",Cn,"posibles combinaciones sin repitición del total de",n,"caracteres", "en grupos de", r,"en",r)
print("Aquí las posibles combinaciones:")
combinaciones_nombres = list(combinations(nombres,r))
for combinations in combinaciones_nombres:
  print(combinations)
```

##Ejemplo 3: Combinaciones con repetición
