```python
import math as mt
from itertools import permutations
```

```python
asistentes = ["as1","as2","as3","as4","as5","as6","as7","as8","as9","as10"]
n = len(asistentes)
r = 4

Pr = mt.factorial(n)/(mt.factorial(n-r))
print("Existen",Pr,"psobles permutaciones sin repetición del total de",n,"nombres, en grupos de",r,"en",r)
print("----------")
print("Aquí las permutaciones:")
permutaciones_nombres = list(permutations(asistentes,r))

for permutacion in permutaciones_nombres:
  print(permutacion)
```
