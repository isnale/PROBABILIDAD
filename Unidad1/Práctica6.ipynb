```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

autos = pd.read_csv("autos2.csv", encoding = "ISO-8859-1")
autos.info()
```

```python
#Estadísticas descriptivas básicas y medidas de tendencia central
print("Valor Mínimo de precios:", autos.price.min())
print("Valor Máximo precios:", autos.price.max())
print("Moda de precios:", autos.price.mode()[0])
print("Media de precios:", autos.price.mean())
print("Mediana de precios:", autos.price.median())
print("Desviacion estandar precios:", autos.price.std())
print("\nDatos estadísticos:\n", autos.price.describe())
```

```python
#Histograma de precios de los automóviles

#Regla de sturges para bins
import math
n = len(autos.price)
k = int(1+math.log2(n))

plt.hist(autos.price, bins=k)
plt.title("Histograma de Precios de Automóviles")
plt.xlabel("Precio")
plt.ylabel("Frecuencia")


```

###Conclusión del gráfico
1. A simple vista se ve como la distribución esta inclinada hacia la izquierda, mostrando como los carros más económicos son los que se venden más. También se aprecia la poca cantidad de autos de lujo , a partir de los $20,000, la cantidad de autos baja drásticamente.
2. Los precios son de un rango variado (5,000-45,000), lo que refleja nuestra desviación estándar de 7947


```python
#Diagrama de cajas y bigotes price-bodystyle
sns.boxplot(x="price", y="body-style", data=autos, hue = "body-style", palette='Set2', legend=False)

plt.title("Distribución según precio y tipo de vehículo ", fontsize=14,fontweight="bold")
plt.xlabel("Precio")
plt.ylabel("Tipo de vehículo")
plt.show()
```

###Conclusión del gráfico:
1. Los hardtop son los autos más caros, seguidos de los convertibles. Por otro lado, los más económicos son los hatchback, con una mediana de precios por debajo de los $10,000.
2. Los sedan es la categoría con más valores atípicos,


```python
#Diagrama de cajas y bigotes price-drive wheels
sns.boxplot(x="price", y="drive-wheels", data=autos, hue = "drive-wheels", palette='Set2', legend=False)

plt.title("Distribución según precio y tipo de ruedas ", fontsize=14,fontweight="bold")
plt.xlabel("Precio")
plt.ylabel("Tipo de ruedas")
plt.legend(title="Tipo de Tracción",labels=["rwd: Tracción trasera","fwd: Tracción delantera","4wd: Tracción 4 ruedas"])
plt.show()
```

###Conclusión del gráfico:
1. Los vehículos de gtracción trasera son los más caros y de una gama más alta, con valores atípicos inclinados hacia la derecha y una caja grande, con una mediana con mayor precio que los otros tipos de tracción.
2. Los autos de tracción delantera son los más baratos, con algunos datos atípicos hasta los $25,000.
3. La tracción de cuatro ruedas y la delantera suelen ser las más accesibles, en comparación con las traseras.



```python
#Diagrama de cajas y bigotes price-engine location
sns.boxplot(x="price", y="engine-location", data=autos, hue = "engine-location", palette='Set2', legend=False)

plt.title("Distribución según precio y lugar del motor ", fontsize=14,fontweight="bold")
plt.xlabel("Precio")
plt.ylabel("Lugar del motor")
plt.show()
```

###Conclusión del gráfico:
Se logra apreciar una clara diferencia entre un motor delantero y uno trasero, siendo este último el de mayor costo.

La mayoría de autos comerciales vienen con el motor delante, abarcando modelos económicos, de gama media, y algunas veces de gama alta de igual manera, pero el motor trasero domina la gama alta, centrandose más en esta, representando un nicho/caja más pequeña y exclusiva
