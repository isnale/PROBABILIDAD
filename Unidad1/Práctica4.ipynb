```python
import random
import matplotlib.pyplot as plt
from statistics import *
```

```python
fabrica1=[]
for i in range (100):
  numero_aleatorio = round(random.uniform(50,100),2)
print (fabrica1)
```

```python
fabrica2=[]
for i in range (100):
  numero_aleatorio2 = round(random.uniform(70,100),2) #Rango de 70 a 100. igual con 2 decimales
  fabrica2.append(numero_aleatorio2)
print (fabrica2)
```

##Diagrama de cajas y bigotes
1. Las líneas se usan para indicar la variabilidad fuera de los cuartiles superior e inferior.
2. Los valores atípicos se representan a veces como puntos individuales que están en línea con los bigotes.
3. **Si los datos son simétricos, la mediana estará en el centro de la caja.**
4. Si los datos están **sesgados**, la mediana estará más cerca de la parte superior o inferior de la caja.
5. Los extremos de arriba y abajo de la caja indican los **cuantiles, o percentiles, 25 y 75.** Estos 2 cuantiles también se conocen como cuartiles, porque separan en cuartos (25%) de los datos. La longitud de la caja es la diferencia entre 2 percentiles y se conoce como rango intercuartílico (IQR).
6. Los diagramas de caja-bigotes son útiles para comparar 2 o + conjuntos de datos.

```python
plt.boxplot(fabrica1)
plt.ylabel('Edades')
plt.title('Producción fábrica 1')
plt.show()
```

```python
#Comparamos ambas producciones
fig, (ax1,ax2) = plt.subplots(nrows=1, ncols=2, figsize=(9, 4))
bplot1=ax1.boxplot(fabrica1, vert=True, patch_artist=True)
ax1.set_title('Producción fábrica 1')

bplot2=ax2.boxplot(fabrica2, vert=True, patch_artist=True)
ax2.set_title('Producción fábrica 2')

plt.show()
```
