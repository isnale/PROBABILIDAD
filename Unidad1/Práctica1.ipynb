##Ejemplo 1: Registro de calificaciones de 12 alumnos


```python
#Lista de calificaciones del grupo A de probabilidad y estadística
GrupoA = [70,70,65,90,100,95,45,80,73,20,100,96]
print(GrupoA)
type(GrupoA) #La función type nos devuelve el tipo de dato de x variable
```

```python
#Sacar la media de las calificaciones con una fórmula
promedio = sum(GrupoA)/len(GrupoA)
print(promedio)
```

```python
#Determinar el valor de la mediana con la función median
mediana = median(GrupoA)
print (mediana)
```

```python
#Determinar el valor de la moda con la función mode

moda = mode(GrupoA)
print(moda)
```

##Ejemplo 3: Histogramas de 2 grupos
1. En este ejemplo generamos 100 números aleatorios reales que representa la producción diaria de 100 telares (metros)
2. Generar otra lista con la producción de 100 telares para una segunda fábrica
3. Con estas listas de valores se determina las medidas de tendencia central: media, moda, mediana, desviación estándar
4. Graficar un histograma para cada fábrica

```python
# Librería para generar números aleatorios
import random
```

```python
fabrica1 = []
for i in range(100):
  num_aleatorio = round(random.uniform(50,100),2)
  fabrica1.append(num_aleatorio)
print(fabrica1)
```

```python
fabrica2 = []
for i in range(100):
  num_aleatorio2 = round(random.uniform(50,100),2)
  fabrica2.append(num_aleatorio2)
print(fabrica2)
```

```python
#Librería de python para graficar
import matplotlib.pyplot as plt
```

```python
plt.hist(fabrica1)
plt.title("Producción diaria de la Fábrica 1")
```

```python

plt.hist(fabrica2)
plt.title("Producción diaria de la Fábrica 2")
```

```python
media = mean(fabrica1)
mediana = median(fabrica1)
moda = mode(fabrica1)

print("Valores de la fábrica 1:")
print("Media:", media)
print("Mediana:", mediana)
print("Moda:", moda)
```

```python
media = mean(fabrica2)
mediana = median(fabrica2)
moda = mode(fabrica2)

print("Valores de la fábrica 2:")
print("Media:", media)
print("Mediana:", mediana)
print("Moda:", moda)
```

```python
#Generar 2 sublots de un reglón por 2 columnas para desplegar los 2 histogramas
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10,5))
ax1.hist(fabrica1)
ax2.hist(fabrica2)
plt.show()
```

```python
# Rangos y frecuencias
counts1, bins1, patches1, = ax1.hist(fabrica1)
print("Limites de Bins Fabrica 1: ", bins1)
print("Frecuencias de Fabrica 1: ", counts1)
print("")
counts2, bins2, patches2, = ax2.hist(fabrica2)
print("Limites de Bins Fabrica 2: ", bins2)
print("Frecuencias de Fabrica 2: ", counts2)
```

```python
media1= mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)
plt.hist(fabrica1, bins=10, alpha=0.2, color='b', edgecolor="black")
plt.axvline(media1, color='b', linestyle='dashed', linewidth=2, label="Media")
plt.axvline(mediana1, color='r', linestyle='dashed', linewidth=2, label="Mediana")
plt.axvline(moda1, color='y', linestyle='dashed', linewidth=2, label="Moda")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.title("Producción diaria de al fábrica 1")
plt.legend()
plt.show()
```

```python
media2= mean(fabrica2)
mediana2 = median(fabrica2)
moda2 = mode(fabrica2)

plt.hist(fabrica2, bins=10, alpha=0.2, color='b', edgecolor="black")
plt.axvline(media2, color='b', linestyle='dashed', linewidth=2, label="Media")
plt.axvline(mediana2, color='r', linestyle='dashed', linewidth=2, label="Mediana")
plt.axvline(moda2, color='y', linestyle='dashed', linewidth=2, label="Moda")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.title("Producción diaria de al fábrica ")
plt.legend()
plt.show()
```
