```python
import random
import matplotlib.pyplot as plt
from statistics import *
```

```python
#Generamos 100 números aleatorios entre 50 y 100
fabrica1 = []
for i in range(100):
  NumeroAleatorio = round(random.uniform(50,100),2)
  fabrica1.append(NumeroAleatorio)
print (fabrica1)
```

```python
#Rango entre 70 y 100
fabrica2 = []
for i in range(100):
  NumeroAleatorio2 = round(random.uniform(70,100),2)
  fabrica2.append(NumeroAleatorio2)
print (fabrica2)
```

```python
maximo1 = max(fabrica1)
minimo1 = min(fabrica1)
rango1 = maximo1 - minimo1
media1 = mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)
DesviacionEstandarFab1 = stdev(fabrica1)
print("Estadístia descriptiva de la fábrica 1 \n", "Media =", media1,
      "\n Mediana =", mediana1, "\n Moda =", moda1, "\n Máximo =", maximo1, "\n Mínimo =", minimo1,
      "\n Desviación estándar =", DesviacionEstandarFab1)
```

```python
maximo2 = max(fabrica2)
minimo2 = min(fabrica2)
rango2 = maximo2 - minimo2

media2 = mean(fabrica2)
mediana2 = median(fabrica2)
moda2 = mode(fabrica2)
DesviacionEstandarFab2 = stdev(fabrica2)
print("Estadística descriptiva de la fábrica 2 \n",
      "Media =", media2,
      "\n Mediana =", mediana2,
      "\n Moda =", moda2,
      "\n Máximo =", maximo2,
      "\n Mínimo =", minimo2,
      "\n Desviación estándar =", DesviacionEstandarFab2)
```

```python
plt.figure(figsize=(8,6))
plt.hist(fabrica1, bins=10, alpha=0.2, color='b', edgecolor = 'black')

plt.axvline(media1, color = 'b', linestyle = 'dashed',linewidth=2, label=f'Media={media1:.2f}')
plt.axvline(mediana1, color = 'r', linestyle = 'dashed',linewidth=2, label=f'Mediana={mediana1:.2f}')
plt.axvline(moda1, color='y', linestyle='dashed', linewidth=2, label=f'Moda={moda1:.2f}')

plt.axvline(media1-DesviacionEstandarFab1, color = 'black', linestyle = 'dashed',linewidth=2, label=f'Desv.Estandár={DesviacionEstandarFab1:.2f}')
plt.axvline(media1+DesviacionEstandarFab1, color = 'black', linestyle = 'dashed',linewidth=2)

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title('Distribución de la fábrica 1')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.hist(fabrica2, bins=10, alpha=0.2, color='b', edgecolor = 'black')

plt.axvline(media2, color = 'b', linestyle = 'dashed',linewidth=2, label=f'Media={media2:.2f}')
plt.axvline(mediana2, color = 'r', linestyle = 'dashed',linewidth=2, label=f'Mediana={mediana2:.2f}')
plt.axvline(moda2, color = 'y', linestyle = 'dashed',linewidth=2, label=f'Moda={moda2:.2f}')

plt.axvline(media2-DesviacionEstandarFab2, color = 'black', linestyle = 'dashed',linewidth=2, label=f'Desv.Estandár={DesviacionEstandarFab2:.2f}')
plt.axvline(media2+DesviacionEstandarFab2, color = 'black', linestyle = 'dashed',linewidth=2)

plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.title('Distribución de la fábrica 2')
plt.legend()
plt.show()
```
