```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
plt.figure(figsize=(6,4))

media=70
desv_std = 4
total_estudiantes=500

peso_minimo = 68
peso_maximo = 72

prob_menos_72 = norm.cdf(peso_maximo,media,desv_std)
prob_menos_68 = norm.cdf(peso_minimo,media,desv_std)

probabilidad = prob_menos_72 - prob_menos_68
estudiantes = total_estudiantes * probabilidad

print("La probabilidad de que un estudiante pese más de 68 y menos de 72 Kg:", probabilidad)
print("Numero de estudiantes que pesan más de 68 y menos de 72 Kg:", estudiantes)

x=np.linspace(media-4*desv_std,media+4*desv_std,1000)
y=norm.pdf(x,media,desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y,where=(x>=peso_minimo)&(x<=peso_maximo),color='skyblue', alpha=0.5)

plt.title('Probabilidad de que un estudiante pese entre 68 y 72 Kg. \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso en Kg de estudiantes')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
