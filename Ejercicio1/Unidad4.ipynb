```python
from scipy.stats import binom
probabilidad_exito = 0.83
num_pacientes = 16
X = 16
probabilidad_ninguno_enfermo = binom.pmf(X,num_pacientes,probabilidad_exito)
print("La probabilidad de que ningún paciente sifra de enfermedad es:", probabilidad_ninguno_enfermo)
```

```python
from scipy.stats import binom
probabilidad_exito = 0.83
num_pacientes = 16
X = 16
probabilidad_todos_enfermos = binom.pmf(X,num_pacientes,probabilidad_exito)
print("La probabilidad de que todos los pacientes sean enfermos es: ", probabilidad_todos_enfermos)
```
