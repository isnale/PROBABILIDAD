3. Calculo de:

   **Valor esperado E(x)**

   **Varianza** y la **desviación estándar**

```python
x = [0,1,2,3,4,5]
p=[0.10,0.25,0.30,0.20,0.10,0.05]

valor_esperado =0
for i in range(len(x)):
  xi = x[i]
  pi = p[i]
  producto = xi * pi
  valor_esperado += producto
  print(f"x = {xi}, p = {pi}, x*p = {producto:.2f}")
print(f"\nValor esperado o Esperanza: {valor_esperado:.2f}")
```

```python
import math
x = [0,1,2,3,4,5]
p=[0.10,0.25,0.30,0.20,0.10,0.05]

esperanza = 0
for i in range(len(x)):
    esperanza += x[i] * p[i]

esperanza_cuadrado = 0
for i in range(len(x)):
    esperanza_cuadrado += (x[i] ** 2) * p[i]

varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

# Mostrar resultados
print(f"E(X) = {esperanza:.2f}")
print(f"E(X^2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
```

4. Construir la **función de distribución acumulada F(x) y representarla gráficamente**

```python
import math
import numpy as np
import matplotlib.pyplot as plt

x = [0,1,2,3,4,5]
p=[0.10,0.25,0.30,0.20,0.10,0.05]

# Calcular función acumulada
F = np.cumsum(p)

# Mostrar resultados
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")

# Graficar
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA)")
plt.xlabel("Calificación")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```
